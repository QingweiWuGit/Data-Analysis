# Yahoo Finance

```python
import time

def download_with_retry(ticker, start, end, attempts=5, delay=30):
    for attempt in range(1, attempts + 1):
        try:
            return yf.download(ticker, start=start, end=end, progress=False)
        except Exception as e:
            msg = str(e)
            if 'YFRateLimitError' in msg or 'rate limited' in msg.lower():
                wait = delay * attempt
                print(f'Rate limited by Yahoo Finance. Retry {attempt}/{attempts} after {wait} seconds...')
                time.sleep(wait)
                continue
            raise
    raise RuntimeError(f'Failed to download {ticker} after {attempts} retries')

df = download_with_retry('AAPL', start='2020-01-01', end='2025-12-31')
df.head()
```

# CoinGecko

```python
from pycoingecko import CoinGeckoAPI
from datetime import datetime
import pandas as pd
import numpy as np
```

```python
cg = CoinGeckoAPI()
```

```python
ohlc = cg.get_coin_ohlc_by_id(
    id='bitcoin', vs_currency='usd', days=30
)
ohlc_df = pd.DataFrame(ohlc, columns=['date', 'open', 'high', 'low', 'close'])
ohlc_df['date'] = pd.to_datetime(ohlc_df['date'], unit='ms')
ohlc_df.set_index('date', inplace=True)
ohlc_df.sort_values(by='date', ascending=False, inplace=True)
ohlc_df.head()
```
